```python
asia_cars = df[df['origin'] == 'Asia']
```

```python
selected_cols = asia_cars[['vehicle_weight', 'model_year']]
```

```python
first_7 = selected_cols.head(7)
```

```python
X = first_7.values
```

```python
import numpy as np
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = (XTX_inv.dot(X.T)).dot(y)
```

```python
print(w.sum())
```
